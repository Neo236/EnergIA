# **Fase 6 — Análisis variables categóricas**

```python
df_energIA.select_dtypes(include=['string[python]']).columns
```

```python
# Variable tipo_inmueble
conteo = df_energIA["tipo_inmueble"].value_counts().reset_index()
conteo.columns = ["tipo_inmueble", "cantidad"]

fig = px.bar(
    conteo,
    x="tipo_inmueble",
    y="cantidad",
    title="Cantidad de hogares por tipo de inmueble",
    color="tipo_inmueble",
    color_discrete_sequence=px.colors.qualitative.Set1
)

fig.update_layout(
    xaxis_title="Tipo de inmueble",
    yaxis_title="Cantidad")

fig.show()
```

```python
# Variable calidad_aislamiento
orden = ["Muy Alta", "Alta", "Media", "Baja", "Muy Baja"]
conteo = df_energIA["calidad_aislamiento"].value_counts().reindex(orden).reset_index()
conteo.columns = ["calidad_aislamiento", "cantidad"]

fig = px.bar(
    conteo,
    x="calidad_aislamiento",
    y="cantidad",
    title="Calidad del aislamiento",
    color="calidad_aislamiento",
    color_discrete_sequence=px.colors.qualitative.Set1
)

fig.update_layout(
    xaxis_title="Calidad del aislamiento",
    yaxis_title="Cantidad",
    xaxis=dict(categoryorder="array",
               categoryarray=orden))

fig.show()
```

```python
# Variable fuente_calefaccion
conteo = df_energIA["fuente_calefaccion"].value_counts().reset_index()
conteo.columns = ["fuente_calefaccion", "cantidad"]

fig = px.bar(
    conteo,
    x="fuente_calefaccion",
    y="cantidad",
    title="Tipos de fuentes de calefacción",
    color="fuente_calefaccion",
    color_discrete_sequence=px.colors.qualitative.Set1
)

fig.update_layout(
    xaxis_title="Tipos de calefacción",
    yaxis_title="Cantidad")

fig.show()
```

```python
# Variable fuente_agua_caliente
conteo = df_energIA["fuente_agua_caliente"].value_counts().reset_index()
conteo.columns = ["fuente_agua_caliente", "cantidad"]

fig = px.bar(
    conteo,
    x="fuente_agua_caliente",
    y="cantidad",
    title="Tipos de calefacción para agua caliente",
    color="fuente_agua_caliente",
    color_discrete_sequence=px.colors.qualitative.Set1
)

fig.update_layout(
    xaxis_title="Tipos de calefacción para agua caliente",
    yaxis_title="Cantidad")

fig.show()
```

```python
# Variable categoria
orden = ["Eficiente", "Moderado", "Ineficiente"]
conteo = df_energIA["categoria"].value_counts().reindex(orden).reset_index()
conteo.columns = ["categoria", "cantidad"]

fig = px.bar(
    conteo,
    x="categoria",
    y="cantidad",
    title="Categoría de eficiencia energética",
    color="categoria",
    color_discrete_sequence=px.colors.qualitative.Set1
)

fig.update_layout(
    xaxis_title="Categoría",
    yaxis_title="Cantidad",
    xaxis=dict(categoryorder="array",
               categoryarray=orden))

fig.show()
```

```python
def boxplot_variable(df, variable, titulo, etiqueta_y):
    fig = px.box(
        df,
        x=variable,
        title=titulo
    )

    fig.update_layout(
        xaxis_title="",
        yaxis_title=etiqueta_y
    )
    fig.update_traces(marker_color="rgb(152,78,163)")
    fig.show()
```

```python
boxplot_variable(
    df_energIA,
    "consumo_kwh",
    "Distribución del consumo eléctrico",
    "Consumo (kWh)"
)
```
